## Paso 1 — Nuestros documentos

```python
documentos=['Brief Manorigen. El proyecto de rotulacion comienza con fotografias, logo, colores, medidas, textos y restricciones. Si la informacion llega dispersa, aparecen correcciones y retrabajo.','Restricciones tecnicas. Puertas, tapa de combustible, molduras, gomas, manillas, emblemas, ventilaciones y patente condicionan la composicion y la instalacion.','Control humano. La IA puede ordenar, comparar y advertir, pero la aprobacion del diseno, las medidas y la viabilidad de instalacion siguen siendo decisiones humanas.']
print(f'{len(documentos)} documentos cargados.')
```

## Paso 2 — Convertir documentos a embeddings

```python
from sentence_transformers import SentenceTransformer
import numpy as np
modelo=SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
embeddings=modelo.encode(documentos)
print(f'Shape: {embeddings.shape}')
```

## Paso 3 — Búsqueda semántica

```python
def buscar(pregunta,top_k=1):
    p_emb=modelo.encode([pregunta])
    sims=np.dot(embeddings,p_emb.T).flatten()
    sims=sims/(np.linalg.norm(embeddings,axis=1)*np.linalg.norm(p_emb))
    top_idx=np.argsort(sims)[::-1][:top_k]
    return [(documentos[i],float(sims[i])) for i in top_idx]
preguntas=['que elementos del vehiculo condicionan la instalacion?','por que se producen retrabajos?','que decisiones debe mantener una persona?']
for q in preguntas:
    print('P:',q)
    print(buscar(q,1))
```
