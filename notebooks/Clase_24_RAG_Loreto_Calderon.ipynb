# Clase 24 — Mini RAG del proyecto
## Agencia de Diseño Gráfico y Comunicación Visual
Este ejercicio bonus simula recuperación de información desde cinco fuentes internas del proyecto antes de responder una pregunta.

```python
documentos = [
 'La agencia enfrenta briefs incompletos, cambios constantes, información dispersa, retrasos y reprocesos.',
 'El Coordinador de Proyectos organiza tareas, plazos y cambios; el Director Creativo valida coherencia con el brief.',
 'La IA puede revisar briefs, detectar faltantes, resumir requerimientos, registrar cambios y generar alertas.',
 'Las decisiones creativas, comerciales y de alto impacto requieren aprobación humana y fuente visible.',
 'TOE y DOI permiten evaluar infraestructura, preparación organizacional, seguridad, compatibilidad, complejidad y adopción.'
]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def recuperar(pregunta, k=2):
    corpus = documentos + [pregunta]
    X = TfidfVectorizer().fit_transform(corpus)
    sims = cosine_similarity(X[-1], X[:-1]).flatten()
    idx = sims.argsort()[::-1][:k]
    return [(documentos[i], round(float(sims[i]),3)) for i in idx]

pregunta = '¿Qué debe hacer la IA y qué decisiones deben seguir siendo humanas?'
recuperar(pregunta)
```

## Respuesta basada en evidencia recuperada
La IA puede apoyar la revisión y organización de información: detectar faltantes, resumir requerimientos, registrar cambios y generar alertas. En cambio, la aprobación creativa, la negociación con clientes y los cambios de alto impacto deben seguir bajo responsabilidad humana.

La idea de RAG es que la respuesta se construya a partir de fuentes del proyecto y no solamente desde la memoria general del modelo.
