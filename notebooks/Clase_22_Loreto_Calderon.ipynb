## Ejercicio 1 — Embeddings con conceptos de rotulación

```python
from sentence_transformers import SentenceTransformer, util
modelo = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
palabras = ['rotulacion','vinilo','impresion','corte','instalacion','vehiculo','branding','logotipo','color','composicion','medidas','cliente']
for p1 in palabras[:6]:
    for p2 in palabras[:6]:
        emb1=modelo.encode(p1,convert_to_tensor=True); emb2=modelo.encode(p2,convert_to_tensor=True)
        print(p1,p2,round(util.cos_sim(emb1,emb2).item(),2))
```

## Ejercicio 2 — Modelo analítico: sentimiento

```python
from transformers import pipeline
clasificador=pipeline('sentiment-analysis',model='nlptown/bert-base-multilingual-uncased-sentiment')
frases=['La propuesta se ve moderna y respeta muy bien la identidad de la marca.','El diseño se corta justo en la tapa de combustible y así no se puede instalar.','La propuesta está bien, pero todavía falta revisar las medidas del portalón.','El cliente aprobó el concepto general, aunque pidió agrandar el logo lateral.','La información llegó incompleta y tuvimos que rehacer parte del montaje.']
for frase in frases: print(frase,clasificador(frase)[0])
```

## Ejercicio 3 — Modelo generativo

```python
generador=pipeline('text-generation',model='datificate/gpt2-small-spanish',max_new_tokens=60)
prompt='En un proyecto de rotulacion vehicular de Manorigen, una buena asistencia de IA deberia'
print(generador(prompt,num_return_sequences=1,do_sample=True,temperature=0.8)[0]['generated_text'])
```
