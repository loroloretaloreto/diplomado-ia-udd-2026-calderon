## Agencia de Diseño Gráfico y Comunicación Visual
Este notebook adapta los ejercicios al asistente IA para gestión y seguimiento de proyectos creativos.

```python
import pandas as pd
mensajes = [
    ['El cliente aprobó la propuesta 2 y podemos preparar originales.', 'aprobacion'],
    ['Falta confirmar medida, material y fecha de entrega.', 'informacion_faltante'],
    ['Cambiar el color secundario y actualizar todas las piezas.', 'solicitud_cambio'],
    ['Necesitamos enviar hoy antes de las 16:00.', 'urgencia'],
    ['Gracias, quedó perfecto.', 'sin_accion']
]
df = pd.DataFrame(mensajes, columns=['mensaje','categoria'])
df
```

### Clasificación simple
El objetivo es representar una función analítica del futuro asistente: distinguir mensajes que requieren acción de mensajes informativos.

```python
def clasificar(texto):
    t = texto.lower()
    if 'aprob' in t: return 'aprobacion'
    if 'falta' in t or 'confirmar' in t: return 'informacion_faltante'
    if 'cambiar' in t or 'modificar' in t: return 'solicitud_cambio'
    if 'hoy' in t or 'urgente' in t: return 'urgencia'
    return 'sin_accion'
df['prediccion_regla'] = df['mensaje'].apply(clasificar)
df
```

### Relaciones semánticas con TF-IDF
Se comparan conceptos propios del flujo de la agencia para observar similitud textual.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
frases = [
 'brief con requerimientos del cliente',
 'información faltante en el brief',
 'solicitud de cambio del cliente',
 'aprobación final del director creativo',
 'problema de producción e impresión'
]
X = TfidfVectorizer().fit_transform(frases)
pd.DataFrame(cosine_similarity(X), index=frases, columns=frases).round(2)
```
