Objetivo: explorar un dataset pequeño y documentar qué decisiones humanas aparecen al crear etiquetas.


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
path=Path('../data/dataset_rotulacion_v1.csv')
if not path.exists(): path=Path('dataset_rotulacion_v1.csv')
df=pd.read_csv(path)
print('Registros:',len(df))
df.head()
```

## 1. Distribución por etapa


```python
print(df['etapa'].value_counts())
df['etapa'].value_counts().plot(kind='bar',title='Registros por etapa')
plt.xlabel('Etapa'); plt.ylabel('Cantidad'); plt.tight_layout(); plt.show()
```

## 2. Urgencia y revisión humana


```python
print(pd.crosstab(df['urgencia'],df['decision_humana_requerida']))
print('Porcentaje con decisión humana requerida:',round(df['decision_humana_requerida'].mean()*100,1),'%')
```

## 3. Riesgos más frecuentes


```python
print(df['riesgo_principal'].value_counts().head(10))
```
