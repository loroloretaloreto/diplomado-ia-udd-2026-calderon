## Agencia de Diseño Gráfico y Comunicación Visual
Dataset sintético para explorar mensajes y eventos de proyectos creativos.

```python
import pandas as pd
url = 'https://raw.githubusercontent.com/loroloretaloreto/diplomado-ia-udd-2026-calderon/main/data/dataset_agencia_proyectos_v1.csv'
df = pd.read_csv(url)
df.head()
```

```python
print('Registros:', len(df))
print('Categorías:', df['categoria'].nunique())
print('Proyectos:', df['tipo_proyecto'].nunique())
df['categoria'].value_counts()
```

```python
pd.crosstab(df['categoria'], df['urgencia'])
```

```python
df.groupby(['rol_responsable','requiere_accion']).size().unstack(fill_value=0)
```

```python
import matplotlib.pyplot as plt
df['categoria'].value_counts().plot(kind='bar')
plt.title('Eventos por categoría')
plt.xlabel('Categoría')
plt.ylabel('Cantidad')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
